## Task 1.2: Train-wise Starting and Ending Stations

```python
df[["Train_No", "Route_Number", "Station_Name"]].head(20)
```

```python
df_sorted = df.sort_values(["Train_No", "Route_Number"])
df_sorted 
```

```python
train_summary = df_sorted.groupby("Train_No").agg(
    Start_Station=("Station_Name", "first"),
    End_Station=("Station_Name", "last")
).reset_index()
train_summary
```

```python
train_summary.head(20)
```

## Task 1.3: Distance and Number of Stops Statistics

```python
df["Distance"].describe()
```

```python
print("Minimum Distance:", df["Distance"].min())
print("Maximum Distance:", df["Distance"].max())
print("Average Distance:", df["Distance"].mean())
print("Median Distance:", df["Distance"].median())
```

```python
stops_per_train = df_sorted.groupby("Train_No").size().reset_index(
    name="Number_of_Stops"
)
```

```python
stops_per_train.head(20)
stops_per_train["Number_of_Stops"].describe()
```
