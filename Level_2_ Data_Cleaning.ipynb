## Task 2.2: Converting Arrival and Departure Times

```python
print("Arrival Time Examples:")
print(df["Arrival_time"].head(20).tolist())

print("\nDeparture Time Examples:")
print(df["Departure_Time"].head(20).tolist())
```

```python
df["Arrival_time"] = pd.to_datetime(
    df["Arrival_time"],
    format="%H:%M:%S",
    errors="coerce"
)

df["Departure_Time"] = pd.to_datetime(
    df["Departure_Time"],
    format="%H:%M:%S",
    errors="coerce"
)
```

```python
df[["Arrival_time", "Departure_Time"]].head()
```

```python
df = df.sort_values(
    ["Train_No", "Route_Number", "SN"]
).reset_index(drop=True)
df
```

```python
train_journey = df.groupby("Train_No").agg(
    Start_Station=("Station_Name", "first"),
    End_Station=("Station_Name", "last"),
    Start_Time=("Departure_Time", "first"),
    End_Time=("Arrival_time", "last"),
    Total_Distance=("Distance", "max"),
    Number_of_Stops=("Station_Name", "count")
).reset_index()
```

```python
train_journey.head(10)
```

```python
print(
    "Arrival 00:00:00:",
    (df["Arrival_time"].dt.strftime("%H:%M:%S") == "00:00:00").sum()
)

print(
    "Departure 00:00:00:",
    (df["Departure_Time"].dt.strftime("%H:%M:%S") == "00:00:00").sum()
)
```

```python
df[
    (df["Arrival_time"].dt.strftime("%H:%M:%S") == "00:00:00") |
    (df["Departure_Time"].dt.strftime("%H:%M:%S") == "00:00:00")
][[
    "Train_No",
    "Route_Number",
    "Station_Name",
    "Arrival_time",
    "Departure_Time",
    "Distance"
]].head(20)
```

```python
df.groupby("Train_No").tail(1)[[
    "Train_No",
    "Station_Name",
    "Arrival_time",
    "Departure_Time",
    "Distance"
]].head(20)
```

```python
df = df.sort_values(
    ["Train_No", "Route_Number", "SN"]
).reset_index(drop=True)
df.head(10)
```

```python
train_journey = df.groupby("Train_No").agg(
    Start_Station=("Station_Name", "first"),
    End_Station=("Station_Name", "last"),
    Start_Time=("Departure_Time", "first"),
    End_Time=("Arrival_time", "last"),
    Total_Distance=("Distance", "max"),
    Number_of_Stops=("Station_Name", "count")
).reset_index()
```

```python
train_journey.head(20)
```

```python
print("Total Trains:", len(train_journey))
```

```python
train_journey.info()
```

```python
train_journey["Start_Time_Calc"] = train_journey["Start_Time"]
train_journey["End_Time_Calc"] = train_journey["End_Time"]
```

```python
train_journey[
    ["Train_No", "Start_Time", "End_Time",
     "Start_Time_Calc", "End_Time_Calc"]
].head(10)
```

```python
train_journey["Journey_Duration"] = (
    train_journey["End_Time_Calc"] -
    train_journey["Start_Time_Calc"]
).dt.total_seconds() / 60
```

```python
train_journey.loc[
    train_journey["Journey_Duration"] < 0,
    "Journey_Duration"
] += 24 * 60   // 24 * 60 means 1,440 minutes = 24 hours.
```

```python
train_journey[
    ["Train_No", "Start_Station", "End_Station",
     "Start_Time", "End_Time",
     "Total_Distance", "Number_of_Stops",
     "Journey_Duration"]
].head(20)
```

```python
print(
    "Journey durations <= 0:",
    (train_journey["Journey_Duration"] <= 0).sum()
)
```

```python
print(
    "Missing journey durations:",
    train_journey["Journey_Duration"].isnull().sum()
)
```

```python
train_journey["Journey_Duration"].describe()
```

```python
train_journey[
    train_journey["Journey_Duration"] > 24 * 60
][[
    "Train_No",
    "Start_Station",
    "End_Station",
    "Start_Time",
    "End_Time",
    "Journey_Duration"
]].head(20)
```

```python
model_data = train_journey[[
    "Train_No",
    "Start_Station",
    "End_Station",
    "Total_Distance",
    "Number_of_Stops",
    "Journey_Duration"
]].copy()
```

```python
model_data.to_csv(
    "train_journey_cleaned.csv",
    index=False
)
```

The train journey dataset was cleaned and transformed from station-level records into a train-level dataset.

The preprocessing steps included checking and handling duplicate records, inspecting missing values, converting arrival and departure times into datetime format, identifying journeys that cross midnight, and calculating total journey duration.

The main machine learning input features created were total distance and number of stops, while journey duration was created as the target variable.

The cleaned train-level dataset was saved as `train_journey_cleaned.csv` for use in the next level.
